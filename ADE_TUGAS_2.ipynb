```python
df = sns.load_dataset("tips")
print(df.head())
```

```python
#Histogram
sns.histplot(df["total_bill"], bins=10, kde=True)
plt.title("Distribusi Total Tagihan Pelanggan")
plt.show()
```

```python
#Boxplot
sns.boxplot(x="sex", y="tip", data=df)
plt.title("Perbandingan Tip Berdasarkan Jenis Kelamin")
plt.show()
```

```python
#Countplot
sns.countplot(x="day", data=df)
plt.title("Jumlah Pelanggan per Hari")
plt.show()
```

```python
#Scatter Plot
sns.scatterplot(x="total_bill", y="tip", data=df)
plt.title("Hubungan Total Tagihan dan Tip")
plt.show()
```

```python
#Heatmap
corr = df.corr(numeric_only=True)
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Matriks Korelasi Variabel Numerik")
plt.show()
```

```python
#Pie Chart
df["sex"].value_counts().plot.pie(autopct="%1.1f%%")
plt.title("Proporsi Pelanggan Berdasarkan Jenis Kelamin")
plt.ylabel("")
plt.show()
```

```python
#Line Chart
data = pd.DataFrame({
    "bulan": pd.date_range("2023-01", periods=6, freq="ME"),
    "penjualan": [210, 250, 300, 290, 310, 350]
})

plt.plot(data["bulan"], data["penjualan"], marker="o")
plt.title("Tren Penjualan per Bulan")
plt.xlabel("Bulan")
plt.ylabel("Jumlah Penjualan")
plt.show()
```
